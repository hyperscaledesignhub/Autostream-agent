### Directory loaders, suitable for multiple text files

```python
from langchain_community.document_loaders import DirectoryLoader
## load all the text files from the directory

dir_loader = DirectoryLoader(
    "data/textfile",
    glob="**/*.txt", #patter to match file,
    loader_cls=TextLoader, #Load the class to use
    show_progress=True
)
documents = dir_loader.load()
print(f" Loaded length of documents {len(documents)}")
for i,doc in enumerate(documents):
    print(f"\n Docuement {i+1}")
    print(f" Source: : {documents[i].metadata}")
    print(f"length : {len(documents[i].page_content)})")
```

```python
### Different text splitting strategies
from langchain.text_splitter import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    TokenTextSplitter
)
print(documents)
```

```python
text = documents[0].page_content
text
```

```python
print("1 Character Text Splitter")
char_splitter = CharacterTextSplitter(
    separator="\n",
    chunk_size = 200,
    chunk_overlap=20,
    length_function=len  
)
char_chunks = char_splitter.split_text(text)
print(f" created ({len(char_chunks)}) chunks")
print(f"First chunk: {char_chunks[0][:100]}")
```

```python
### Lets see multiple chunks its been splitted into
print("printing all chunks")
print(char_chunks)
print(char_chunks[0])
print(char_chunks[1])
print(char_chunks[2])
```

```python
# Method 2 Recursive character Text splitter 
print("\n RECURSIVE CHARACTER TEXT SPLITTER")
recursive_splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n","\n","",""], #Try these separators in order 
    chunk_size=200,
    chunk_overlap=20,
    length_function=len
)

recursive_chunks = recursive_splitter.split_text(text)
print(f"Created {len(recursive_chunks)} chunks")
print(f"First chunk: {recursive_chunks[0][:100]}")

```

```python
# Method 2 Recursive character Text splitter, by using only empty
# separator we can see there is overalp across 2 chunks
print("\n RECURSIVE CHARACTER TEXT SPLITTER")
recursive_splitter = RecursiveCharacterTextSplitter(
    separators=[" "], #Using only empty separator 
    chunk_size=200,
    chunk_overlap=20,
    length_function=len
)

recursive_chunks = recursive_splitter.split_text(text)
print(f"Created {len(recursive_chunks)} chunks")
print(f"chunk-0:  {recursive_chunks[0]}")
print("--------------------------------")
print(f"chunk-1:  {recursive_chunks[1]}")
```

```python
##Token based splitter, this is going to count space and all characters
print("\n TOKEN TEXT SPLITTER")
token_splitter = TokenTextSplitter(
    chunk_size=50,
    chunk_overlap=10
)
token_chunks=token_splitter.split_text(text)
print(f"Created {len(token_chunks)} chunks")
print(f"chunk-0:  {token_chunks[0]}")
print("--------------------------------")
print(f"chunk-1:  {token_chunks[1]}")

```
